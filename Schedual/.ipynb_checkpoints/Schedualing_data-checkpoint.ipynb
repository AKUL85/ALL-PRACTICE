```python
# ==============================
# Updated Scheduler for Your CSV Header
# ==============================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import requests

# -----------------------------
# CONFIG
# -----------------------------
API_KEY = "lX61kI4GCdQ3Dn80OXvI"  # ElectricityMap API key
REGION = "US-CA"                 # India, Maharashtra

# Thermal & reliability parameters
T_MAX = 80      # Max safe temperature (Celsius)
ALPHA = 0.8
BETA = 0.05
GAMMA = 0.1
KAPPA = 0.05
LAMBDA_THERMAL = 0.5
LAMBDA_RELIABILITY = 0.5

# -----------------------------
# LOAD Alibaba Workload Trace
# -----------------------------
workload_file = "alibaba_sample.csv"
df = pd.read_csv(workload_file)
print("Workload data loaded:", df.shape)

# Convert CPU milli to cores
df['cpu'] = df['cpu_milli'] / 1000

# Convert scheduled_time to numeric/ordinal for sorting (seconds since epoch)
df['time'] = pd.to_datetime(df['scheduled_time'])
df['time'] = df['time'].astype(np.int64) // 10**9  # seconds
df = df.sort_values('time')

# Use only necessary columns
workloads = df[['time', 'name', 'cpu']].rename(columns={'name':'workload_id'})
print("Processed workloads:", workloads.head())

# -----------------------------
# FETCH Carbon Intensity
# -----------------------------
def fetch_carbon(region):
    url = f"https://api.electricitymap.org/v3/carbon-intensity/latest?zone={region}"
    headers = {"auth-token": API_KEY}
    response = requests.get(url, headers=headers)
    if response.status_code != 200:
        raise Exception(f"Failed to fetch data: {response.text}")
    data = response.json()
    intensity = data['carbonIntensity']  # gCO2/kWh
    return intensity

carbon_intensity = fetch_carbon(REGION)
print("Carbon intensity fetched:", carbon_intensity, "gCO2/kWh")

# -----------------------------
# SIMULATION FUNCTION
# -----------------------------
def run_scheduler(workloads, sites, mode="integrated"):
    site_temps = {s: 25.0 for s in sites}
    assignment = {}
    total_carbon = 0
    max_temp = 0
    total_reliability_penalty = 0

    for t in sorted(workloads['time'].unique()):
        w_t = workloads[workloads['time']==t]
        for _, w in w_t.iterrows():
            best_score = float('inf')
            best_site = None
            for s in sites:
                temp_next = ALPHA*site_temps[s] + BETA*w['cpu'] + GAMMA*25
                thermal_penalty = max(0, temp_next - T_MAX)
                reliability_penalty = np.exp(KAPPA * thermal_penalty)
                carbon_cost = w['cpu'] * carbon_intensity / 1000  # kWh assumed

                # Score based on mode
                if mode=="carbon":
                    score = carbon_cost
                elif mode=="thermal":
                    score = thermal_penalty
                elif mode=="reliability":
                    score = reliability_penalty
                else:
                    score = carbon_cost + LAMBDA_THERMAL*thermal_penalty + LAMBDA_RELIABILITY*reliability_penalty

                if score < best_score:
                    best_score = score
                    best_site = s
                    best_temp_next = temp_next

            assignment[(w['workload_id'], t)] = best_site
            site_temps[best_site] = best_temp_next
            total_carbon += carbon_cost
            max_temp = max(max_temp, best_temp_next)
            total_reliability_penalty += reliability_penalty

    metrics = {
        "Total Carbon (kg CO2)": total_carbon,
        "Max Temperature (C)": max_temp,
        "Total Reliability Penalty": total_reliability_penalty
    }
    return assignment, metrics

# -----------------------------
# RUN BASELINES + INTEGRATED
# -----------------------------
sites = ["SiteA", "SiteB", "SiteC"]
modes = ["carbon", "thermal", "reliability", "integrated"]
results = {}

for mode in modes:
    _, metrics = run_scheduler(workloads, sites, mode=mode)
    results[mode] = metrics
    print(f"\nMode: {mode}")
    for k,v in metrics.items():
        print(f"{k}: {v:.2f}")

# -----------------------------
# PLOT COMPARISON
# -----------------------------
df_results = pd.DataFrame(results).T
df_results.plot(kind='bar', figsize=(10,6))
plt.title("Baseline Comparison: Scheduler Metrics")
plt.ylabel("Value")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# -----------------------------
# WORKLOAD DISTRIBUTION PLOT (INTEGRATED)
# -----------------------------
assignment_integrated, _ = run_scheduler(workloads, sites, mode="integrated")
workload_counts = pd.Series(list(assignment_integrated.values())).value_counts()
plt.bar(workload_counts.index, workload_counts.values)
plt.title("Workload Distribution Across Sites (Integrated Scheduler)")
plt.ylabel("Number of Workloads")
plt.show()
```
